```python
%python
# Databricks Notebook: 01_create_bronze_delta_tables
# Purpose: Write all pharma datasets as Delta tables to Bronze container

from pyspark.sql.functions import current_timestamp, lit
from datetime import datetime

print("="*60)
print("CREATING BRONZE DELTA TABLES FOR ALL PHARMA DATASETS")
print("="*60)

# Storage account configuration
storage_account = "pharmadatalake7"
raw_base_path = f"abfss://raw@{storage_account}.dfs.core.windows.net/"
bronze_base_path = f"abfss://bronze@{storage_account}.dfs.core.windows.net/"

# Read and Write each table as Delta
print("\nReading raw CSV files and writing as Delta tables...\n")

# 1. Clinical Trials Table
print("Processing Clinical Trials...")
df_clinical_trials = spark.read.option("header", True).option("inferSchema", True).csv(f"{raw_base_path}pharma_1_clinical_trials.csv")
df_clinical_trials_with_metadata = df_clinical_trials \
    .withColumn("ingestion_timestamp", current_timestamp()) \
    .withColumn("source_file", lit("pharma_1_clinical_trials.csv")) \
    .withColumn("ingestion_batch_id", lit(datetime.now().strftime("%Y%m%d_%H%M%S")))
df_clinical_trials_with_metadata.write.format("delta").mode("overwrite").save(f"{bronze_base_path}clinical_trials/")
print(f"✓ Clinical Trials Delta table created")

# 2. Sales Table
print("Processing Sales...")
df_sales = spark.read.option("header", True).option("inferSchema", True).csv(f"{raw_base_path}pharma_2_sales_revenue.csv")
df_sales_with_metadata = df_sales \
    .withColumn("ingestion_timestamp", current_timestamp()) \
    .withColumn("source_file", lit("pharma_2_sales_revenue.csv")) \
    .withColumn("ingestion_batch_id", lit(datetime.now().strftime("%Y%m%d_%H%M%S")))
df_sales_with_metadata.write.format("delta").mode("overwrite").save(f"{bronze_base_path}sales/")
print(f"✓ Sales Delta table created")

# 3. Inventory Table
print("Processing Inventory...")
df_inventory = spark.read.option("header", True).option("inferSchema", True).csv(f"{raw_base_path}pharma_3_inventory_supply_chain.csv")
df_inventory_with_metadata = df_inventory \
    .withColumn("ingestion_timestamp", current_timestamp()) \
    .withColumn("source_file", lit("pharma_3_inventory_supply_chain.csv")) \
    .withColumn("ingestion_batch_id", lit(datetime.now().strftime("%Y%m%d_%H%M%S")))
df_inventory_with_metadata.write.format("delta").mode("overwrite").save(f"{bronze_base_path}inventory/")
print(f"✓ Inventory Delta table created")

# 4. Prescriptions Table
print("Processing Prescriptions...")
df_prescriptions = spark.read.option("header", True).option("inferSchema", True).csv(f"{raw_base_path}pharma_4_prescriptions.csv")
df_prescriptions_with_metadata = df_prescriptions \
    .withColumn("ingestion_timestamp", current_timestamp()) \
    .withColumn("source_file", lit("pharma_4_prescriptions.csv")) \
    .withColumn("ingestion_batch_id", lit(datetime.now().strftime("%Y%m%d_%H%M%S")))
df_prescriptions_with_metadata.write.format("delta").mode("overwrite").save(f"{bronze_base_path}prescriptions/")
print(f"✓ Prescriptions Delta table created")

# 5. Side Effects Table
print("Processing Side Effects...")
df_side_effects = spark.read.option("header", True).option("inferSchema", True).csv(f"{raw_base_path}pharma_5_side_effects.csv")
df_side_effects_with_metadata = df_side_effects \
    .withColumn("ingestion_timestamp", current_timestamp()) \
    .withColumn("source_file", lit("pharma_5_side_effects.csv")) \
    .withColumn("ingestion_batch_id", lit(datetime.now().strftime("%Y%m%d_%H%M%S")))
df_side_effects_with_metadata.write.format("delta").mode("overwrite").save(f"{bronze_base_path}side_effects/")
print(f"✓ Side Effects Delta table created")

# 6. R&D Pipeline Table
print("Processing R&D Pipeline...")
df_rd_pipeline = spark.read.option("header", True).option("inferSchema", True).csv(f"{raw_base_path}pharma_6_rd_pipeline.csv")
df_rd_pipeline_with_metadata = df_rd_pipeline \
    .withColumn("ingestion_timestamp", current_timestamp()) \
    .withColumn("source_file", lit("pharma_6_rd_pipeline.csv")) \
    .withColumn("ingestion_batch_id", lit(datetime.now().strftime("%Y%m%d_%H%M%S")))
df_rd_pipeline_with_metadata.write.format("delta").mode("overwrite").save(f"{bronze_base_path}rd_pipeline/")
print(f"✓ R&D Pipeline Delta table created")

# 7. Pricing Table
print("Processing Pricing...")
df_pricing = spark.read.option("header", True).option("inferSchema", True).csv(f"{raw_base_path}pharma_7_pricing_reimbursement.csv")
df_pricing_with_metadata = df_pricing \
    .withColumn("ingestion_timestamp", current_timestamp()) \
    .withColumn("source_file", lit("pharma_7_pricing_reimbursement.csv")) \
    .withColumn("ingestion_batch_id", lit(datetime.now().strftime("%Y%m%d_%H%M%S")))
df_pricing_with_metadata.write.format("delta").mode("overwrite").save(f"{bronze_base_path}pricing/")
print(f"✓ Pricing Delta table created")

# 8. Physician Prescribing Table
print("Processing Physician Prescribing...")
df_physician = spark.read.option("header", True).option("inferSchema", True).csv(f"{raw_base_path}pharma_8_physician_prescribing.csv")
df_physician_with_metadata = df_physician \
    .withColumn("ingestion_timestamp", current_timestamp()) \
    .withColumn("source_file", lit("pharma_8_physician_prescribing.csv")) \
    .withColumn("ingestion_batch_id", lit(datetime.now().strftime("%Y%m%d_%H%M%S")))
df_physician_with_metadata.write.format("delta").mode("overwrite").save(f"{bronze_base_path}physician/")
print(f"✓ Physician Prescribing Delta table created")

# 9. Marketing Table
print("Processing Marketing...")
df_marketing = spark.read.option("header", True).option("inferSchema", True).csv(f"{raw_base_path}pharma_9_marketing_promotions.csv")
df_marketing_with_metadata = df_marketing \
    .withColumn("ingestion_timestamp", current_timestamp()) \
    .withColumn("source_file", lit("pharma_9_marketing_promotions.csv")) \
    .withColumn("ingestion_batch_id", lit(datetime.now().strftime("%Y%m%d_%H%M%S")))
df_marketing_with_metadata.write.format("delta").mode("overwrite").save(f"{bronze_base_path}marketing/")
print(f"✓ Marketing Delta table created")


print("\n" + "="*60)
print("✅ ALL BRONZE DELTA TABLES CREATED SUCCESSFULLY!")
print("="*60)
```
